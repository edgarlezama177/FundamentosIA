```python
import matplotlib.pyplot as plt
import networkx as nx
```

```python
grafo = nx.Graph()
```

```python
grafo.add_edges_from(
    [
        ("Servidor_1", "Router A"),
        ("Router A", "Switch A"),
        ("Switch A", "PC1"),
        ("Router A", "Switch B"),
        ("Switch A", "Impresor A"),
        ("Impresor A", "Switch A"),
        ("Switch B", "Impresor A"),
        ("Switch B", "PC 2")
    ] 
)
```

```python
nx.draw(
    grafo,
    with_labels=True,
    node_size=1000
)
plt.show
```
