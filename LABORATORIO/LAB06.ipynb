```python
#Librería utilizada para graficar
import matplotlib.pyplot as plt
#Librería que permite trabajar con estructuras complejas como Grafos
import networkx as nx
```

```python
grafo = nx.Graph()
```

```python
grafo.add_edges_from(
    [
        ("A", "B"),
        ("B", "D"),
        ("D", "F"),
        ("F", "H"),
        ("H", "I"),
        ("I", "J"),
        ("A", "C"),
        ("C", "G"),
        ("G", "E"),
        ("E", "F"),
        ("G", "J")
    ] 
)
```

```python
nx.draw(
    grafo,
    with_labels=True,
    node_size=1000
)
plt.show
```

```python
grafo.remove_edge ("G", "J")
```

```python
#Mostrar rutas

rutas=list(
    nx.all_simple_paths(grafo, source="A", target="J")
)
```

```python
#Imprimir rutas
print("Rutas encontradas")

for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
#Contar pasos

pasos1 = len(rutas[0])-1
print("Pasos de la ruta 1: ", pasos1)
pasos2 = len(rutas[1])-1
print("Pasos de la ruta 2: ", pasos2)
```

```python
#Proposiciones

P = pasos1 < pasos2 
Q = pasos2 < pasos1 
if P:
    print("P=True")
    print("La ruta 1 es la mejor opción")
elif Q:
    print("Q=True")
    print("La ruta 2 es la mejor opción")
else:
    print("Las rutas son iguales")
```
