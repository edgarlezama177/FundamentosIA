```python
# Situación: Un agente inteligente se encuentra en A y necesita llegar a G.
```

```python
#Librería utilizada para graficar
import matplotlib.pyplot as plt
#Librería que permite trabajar con estructuras complejas como Grafos
import networkx as nx
```

```python
#Creamos una instancia de la clase Graph
#Graph es una colección de nodos

grafo = nx.Graph() 
```

```python
#Creamos el grafo utilizando el método add_edges

grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("A","E"),
        ("B","D"),
        ("D","F"),
        ("E","G"),
        ("E","F")
    

    ]
)
```

```python
#Mostrar el grafo

nx.draw(
    grafo,
    with_labels = True,
    node_size=1000
)
plt.show
```

```python
#Mostrar las rutas posibles del grafo

rutas = list(
    nx.all_simple_paths(grafo,source="A", target="G")
)
```

```python
#Imprimir las rutas
print("Rutas encontradas")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
#Mostrar el total de pasos de cada ruta
pasos1 = len(rutas[0])-1
print("Pasos de la ruta 1: ", pasos1)
pasos2 = len(rutas[1])-1
print("Pasos de la ruta 2: ", pasos2)
```

```python
#Aplicamos la logica proposicional

P = pasos1 < pasos2 
Q = pasos2 < pasos1

if P:
    print("P = True")
    print("La ruta 1 es mejor opción.")
elif Q:
    print("Q = True")
    print("La ruta 2 es mejor opción.")
else:
    print("P = False")
    print("Q = False")
    print("Las dos rutas son iguales, haz lo que quieras!")
```
